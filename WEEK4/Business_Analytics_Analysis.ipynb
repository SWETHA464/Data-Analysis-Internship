```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file_path = 'Sample data (1).xlsx'
df = pd.read_excel(file_path)
df.columns = [c.strip() for c in df.columns]
df.head()
```

```python
print(df.shape)
print(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
df['Discount Band'] = df['Discount Band'].fillna('Unknown')
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df = df.drop_duplicates().copy()
print(df.info())
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
margin = total_profit / total_sales
print(f'Sales: ${total_sales:,.2f}')
print(f'Profit: ${total_profit:,.2f}')
print(f'Profit Margin: {margin:.2%}')
print(f'Units Sold: {df["Units Sold"].sum():,.0f}')
```

```python
def summary(dim):
    out = df.groupby(dim).agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Units=('Units Sold','sum'), Discounts=('Discounts','sum'))
    out['Margin %'] = out['Profit']/out['Sales']
    return out.sort_values('Sales', ascending=False)

segment = summary('Segment'); country = summary('Country'); product = summary('Product'); discount = summary('Discount Band')
segment, country, product, discount
```

```python
monthly = df.groupby(['Year','Month Number','Month Name']).agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Units=('Units Sold','sum')).reset_index().sort_values(['Year','Month Number'])
monthly['Margin %'] = monthly['Profit']/monthly['Sales']
monthly
```

```python
monthly.plot(x='Month Name', y='Sales', kind='line', marker='o', title='Monthly Sales Trend')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
segment[['Sales','Profit']].plot(kind='bar', title='Sales and Profit by Segment')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.to_csv('PowerBI_Cleaned_Data.csv', index=False)
monthly.to_csv('monthly_summary.csv', index=False)
print('Exports created.')
```
