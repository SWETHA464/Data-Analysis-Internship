```python
import pandas as pd
import sqlite3

df = pd.read_excel('/content/SQL_Sales_Dataset_200_Rows.xlsx')

df.head()
```

```python
conn = sqlite3.connect('sales.db')

df.to_sql('sales', conn, if_exists='replace', index=False)

print("Sales table created successfully.")
```

```python
query = """
SELECT *
FROM sales
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT order_id, customer_name, category, total_price
FROM sales
WHERE total_price > 20000;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT category, SUM(total_price) AS total_revenue
FROM sales
GROUP BY category;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT order_id, customer_name, category, total_price
FROM sales
ORDER BY total_price DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT SUM(total_price) AS total_revenue
FROM sales;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT ROUND(AVG(total_price), 2) AS average_order_value
FROM sales;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT COUNT(order_id) AS total_orders
FROM sales;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT category, COUNT(order_id) AS number_of_orders
FROM sales
GROUP BY category
ORDER BY number_of_orders DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    category,
    COUNT(order_id) AS number_of_orders,
    SUM(total_price) AS total_revenue,
    ROUND(AVG(total_price), 2) AS average_order_value
FROM sales
GROUP BY category
ORDER BY total_revenue DESC;
"""

pd.read_sql_query(query, conn)
```

```python
region_data = pd.DataFrame({
    'region': ['North', 'South', 'East', 'West'],
    'region_manager': ['Manager A', 'Manager B', 'Manager C', 'Manager D']
})

region_data.to_sql('regions', conn, if_exists='replace', index=False)
```

```python
query = """
SELECT
    s.order_id,
    s.customer_name,
    s.region,
    r.region_manager,
    s.total_price
FROM sales s
INNER JOIN regions r
    ON s.region = r.region;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    customer_name,
    SUM(total_price) AS total_spent
FROM sales
GROUP BY customer_name
HAVING SUM(total_price) > (
    SELECT AVG(customer_total)
    FROM (
        SELECT SUM(total_price) AS customer_total
        FROM sales
        GROUP BY customer_name
    )
)
ORDER BY total_spent DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    order_id,
    customer_name,
    total_price,
    CASE
        WHEN total_price >= 20000 THEN 'High Value'
        WHEN total_price >= 10000 THEN 'Medium Value'
        ELSE 'Low Value'
    END AS order_category
FROM sales
ORDER BY total_price DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    customer_name,
    SUM(total_price) AS total_spent,
    COUNT(order_id) AS total_orders
FROM sales
GROUP BY customer_name
ORDER BY total_spent DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT
    ROUND(AVG(total_price), 2) AS average_order_value
FROM sales;
"""

pd.read_sql_query(query, conn)
```
