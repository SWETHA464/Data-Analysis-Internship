```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/data.csv")
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df.dtypes
```

```python
df["Year"] = df["Date"].dt.year.astype("Int64")
df[["Date", "Year"]].head()
```

```python
df["Heart_Rate_Range"] = df["Maxpulse"] - df["Pulse"]
df[["Pulse", "Maxpulse", "Heart_Rate_Range"]].head()
```

```python
df["Intensity"] = pd.cut(
    df["Pulse"],
    bins=[0, 99, 129, 159, float("inf")],
    labels=["Low", "Moderate", "High", "Very High"]
)
df[["Pulse", "Intensity"]].head(10)
df["Intensity"].value_counts()
```

```python
df["Calories_per_Minute"] = df["Calories"] / df["Duration"]
df[["Duration", "Calories", "Calories_per_Minute"]].head()
```

```python
df["Workout_Level"] = pd.cut(
    df["Calories"],
    bins=[0, 300, 500, 700, float("inf")],
    labels=["Low", "Moderate", "High", "Very High"]
)
df[["Calories", "Workout_Level"]].head(10)
```

```python
numeric_columns = df.select_dtypes(include="number").columns

numeric_columns
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
df.isnull().sum()
```

```python
print("Mean:", df["Calories"].mean())
print("Median:", df["Calories"].median())
print("Mode:", df["Calories"].mode()[0])
print("Variance:", df["Calories"].var())
```
